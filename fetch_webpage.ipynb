```python
import requests
from urllib.parse import quote

title = "Reconstruction:Proto-West_Germanic/fallijan"
encoded_title = quote(title, safe=":/")

url = "https://en.wiktionary.org/w/api.php"
params = {
    "action": "query",
    "format": "json",
    "titles": encoded_title,
    "prop": "extracts",
    "explaintext": True
}
headers = {"User-Agent": "ProtoWordFetcher/1.0 (contact@example.com)"}

response = requests.get(url, params=params, headers=headers)

if response.status_code == 200:
    data = response.json()
    pages = data["query"]["pages"]
    for _, p in pages.items():
        print("Title:", p.get("title"))
        print("Exists:", "missing" not in p)
        print(p.get("extract", "")[:800])
else:
    print("HTTP Error:", response.status_code)
    print(response.text)
```

```python
import wikipediaapi

wiktionary = wikipediaapi.Wikipedia(
    user_agent="ProtoWordFetcher/1.0 (contact@example.com)",
    language='en',
    extract_format=wikipediaapi.ExtractFormat.WIKI,
)

wiktionary._wiki_base_url = 'https://en.wiktionary.org'
wiktionary._api_url = 'https://en.wiktionary.org/w/api.php'

title = "Reconstruction:Proto-West_Germanic/fallijan"
page = wiktionary.page(title)

print(page.title)
print(page.exists())
print(page.text[:800])
```
